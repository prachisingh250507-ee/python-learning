```python
#Program to Check whether a number is a palindrome.
num=int(input("Enter a number: "))
num_str=str(num)
if num_str==num_str[::-1]:
    print(f"The given number {num} is a palindrome!")
else:
    print(f"The given number {num} is not a palindrome!")
```

```python
# Check whether a number is an Armstrong number.
import math as m
num=int(input("Enter a number: "))
num_str=str(num)
sum=0
d=len(num_str)
print(f"No. of digits in {num} are: {d}")

for i in range(d):
  if (d-i)>0 and num>0:
    f=num%10
    p=m.pow(f,d)
    sum=int(sum+p)
    num=num//10

sum_str=str(sum)
print(f"Sum of {num_str} is {sum_str}")

if num_str==sum_str:
  print("Therefore, Given no. is an Armstrong number")
else:
  print("Given no. is not an Armstrong number")
```

```python
# Print all Armstrong numbers from 1 to 1000.
print("armstrong numbers between 1 to 1000 are: ")
for i in range(1,1000):
  i_str=str(i)
  digits=len(i_str)
  sum=0
  for j in range(digits):
    sum+=int(i_str[j])**digits
  if sum==i:
    print(i)
```

```python
# Print Fibonacci series up to N terms.

def fibonacci_series(n):
    if n <= 0:
        print("Please enter a positive integer for the number of terms.")
        return
    elif n == 1:
        print("Fibonacci Series:")
        print(0)
        return

    a, b = 0, 1
    print("Fibonacci Series:")
    print(a)
    print(b)

    for i in range(2, n):
        nxt_term = a + b
        print(nxt_term)
        a = b
        b = nxt_term

try:
    N = int(input("Enter the number of terms (N) for the Fibonacci series: "))
    fibonacci_series(N)
except ValueError:
    print("Invalid input. Please enter an integer.")
```

```python
# Check whether a number is prime.
num=int(input("Enter a number:"))
if 0<=num<2:
  print(f"No, {num} is not a prime number!")
elif num==2:
  print("Yes, two is a prime number!")
for i in range(2,num):
  if num%i==0:
    print(f"No, {num} is not a prime number!")
  else:
    print(f"Yes, {num} is a prime number!")
```
