```python
# Print all prime numbers between 1 and 100.
print("Prime Numbers between 1 and 100 are:")

for i in range(2, 101):
    if i in (2, 3, 5, 7):
        print(i, end=" ")
    elif i % 2 != 0 and i % 3 != 0 and i % 5 != 0 and i % 7 != 0:
        print(i, end=" ")
```

```python
# Pattern printing 1
for i in range (1,7) :
  pattern = "*"
  print(pattern*i)
```

```python
# Pattern printing 2
for i in range (6,0,-1) :
  pattern = "*"
  print(pattern*i)
```

```python
# Pattern printing 3
for i in range (1,6):
  for j in range (1,i+1):
    print(j, end="")
  print()
```

```python
# Pattern printing 4
for i in range (1,6) :
  for j in range (1,i+1):
    print(i, end="")
  print()
```

```python
# Pattern printing 5
lst1= ["A","B","C","D","E"]
for i in lst1 :
  for j in lst1[0:lst1.index(i)+1]:
    print(j, end="")
  print()
```

```python
# Pattern printing 6
#    *
#   ***
#  *****
# *******
#*********

for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end="")
    for k in range(2 * i - 1):
        print("*", end="")
    print()

```
